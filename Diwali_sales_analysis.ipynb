```python
import pandas as pd
df= pd.read_csv('/content/drive/MyDrive/DiwaliData.csv', encoding='latin1')
df.head(5)
```

```python
df.shape
```

```python
df.info()
```

```python
df.drop(['Status','unnamed1'],axis=1, inplace= True)
```

```python
df.info()
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df['Amount'] = df['Amount'].fillna(df['Amount'].mean())
```

```python
df.shape
```

```python
df.info()
```

```python
df['Amount']= df['Amount'].astype('int')
```

```python
df['Amount'].dtype
```

```python
df.columns
```

```python
# rename the column
df.rename(columns= {'Marital_Status':'Shaadi'})
```

```python
df.columns
```

```python
import seaborn as sns
ax= sns.countplot(x='Gender', data=df)

for bars in ax.containers:
  ax.bar_label(bars)
```

```python
sales_gen= df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x= 'Amount', y= 'Gender', data=sales_gen)
```

From the above groaph we can see that most of the buyers are female and even the purchasing power of Female is greater then men.

## **Age**

```python
df.columns
```

```python
ad=sns.countplot(x='Age Group', data=df)

for bars in ad.containers:
  ad.bar_label(bars)
```

```python
ax= sns.countplot(x='Age Group', hue='Gender', data=df)

for bars in ax.containers:
  ax.bar_label(bars)
```

From the above graph people with the age group of 26-35 are highest buyers and in the female bought then male.

```python
sales_age= df.groupby(['Age Group'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x= 'Age Group', y= 'Amount', data=sales_age)
```

## State

```python
# total number of orders from top 10 states
sales_state= df.groupby(['State'], as_index= False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)
sns.set(rc={'figure.figsize':(17,5)})
sns.barplot(data=sales_state, x='State', y='Orders')
```

```python
# Total amount/ sales from top 10 states
sales_state= df.groupby(['State'], as_index= False)['Amount'].sum().sort_values(by='Amount', ascending= False) #.head(10)
sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data=sales_state, x='State', y='Amount')

ax= sns.barplot(data=sales_state, x='State', y='Amount')
for bars in ax.containers:
  ax.bar_label(bars)
```

From the above graph we can see that most of the orders are from Uttar pradesh, Maharashtra and Karnakata respectively.
